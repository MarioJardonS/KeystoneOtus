```python
plantas = ["Arachis_hypogaea" , "Capsicum_annuum" , "Glycine_max" , "Lactuca_sativa_L" , "Malus_domestica" , "Oryza_sativa" , "Panax_ginseng" , "Panax_notoginseng" , "Persea_americana_Mill" , "Sorghum_bicolor" , "Triticum_aestivum" , "Zea_mays_L._hybrid"]
```

```python
import pandas as pd

keystone = []
for i in range(len(plantas)):
    file_i = pd.read_csv(plantas[i] + "_Rhizosphere_keystone_otus_taxonomy.csv"  )
    file_i = file_i[["OTU" , "Genus" , "Species"]]
    keystone.append(file_i)
```

```python
for file in keystone:
    print(file.head())
```

```python
file = open("../../../../ncbi_scraper.txt")
```

```python
def palabras_especie(linea):
    mots = []
    i = 1
    mot = ""
    while linea[i] != "}":
        if linea[i] == ",":
            mots.append(mot)
            mot = ""
        elif linea[i] != " " and linea[i] != "'":
            mot = mot + linea[i]
        
        i = i + 1 
        
    mots.append(mot)
    
    mots.remove('rhizosphere')
    if mots[0][0].isupper() == False:
        mots.reverse()
    
    if len(mots)  > 1:
        mots = mots[0] + "_" + mots[1]        
    
    return mots
        
```

```python
especies = set()

for line in file:
    if line[0] == "{":
        especie = palabras_especie(line)
        if len(especie) > 1:
            
            especies.add(especie)
            
                
```

```python
especies
```

```python
def del_(st):
    i = 0 
    while st[i] != "_":
        i = i + 1
        
    return [st[:i] , st[i+1:]]
```

```python
for especie in especies: 
    #print(especie)
    especie = del_(especie)
    #print(especie)
    presencias = []
    for j in range(len(keystone)):
        #print(keystone[j].shape[0])
        for i in range(keystone[j].shape[0]):
            if especie[0] == keystone[j].loc[i,["Genus" , "Species"]][0] and especie[1] == keystone[j].loc[i,["Genus" , "Species"]][1]:
                presencias.append(1)
                break
                
                
                
        else:
            presencias.append(0)
            
            
    print(presencias)
        
```
